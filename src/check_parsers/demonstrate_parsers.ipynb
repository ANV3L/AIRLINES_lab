```python
# Демонстрация работы парсеров

В этом ноутбуке показано соответствие исходных (raw) данных и результата их парсинга в единый CSV-формат. 
Для каждого формата данных выводятся последние строки исходного файла и результата парсинга.
```

```python
import pandas as pd
from IPython.display import display, HTML

RAW_YAML = "resources/raw/SkyTeam-Exchange.yaml"
PARSED_YAML = "resources/processed/yaml_parsed.csv"

with open(RAW_YAML, 'r', encoding='utf-8') as f:
    all_lines = f.readlines()
    raw_text = "".join(all_lines[-10:])  

df_yaml = pd.read_csv(PARSED_YAML, sep='|').tail(10)

html_yaml = f"""
<div style="display: flex; justify-content: space-between; gap: 20px;">
    <div style="width: 45%; background: #f8f9fa; padding: 15px; border-radius: 8px; border: 1px solid #dee2e6;">
        <h4 style="margin-top: 0; color: #495057;"> Было: Raw YAML </h4>
        <pre style="font-family: monospace; font-size: 13px; color: #d63384; overflow-x: auto; margin: 0;">{raw_text}</pre>
    </div>
    <div style="width: 55%;">
        <h4 style="margin-top: 0; color: #198754;"> Стало: Parsed CSV </h4>
        {df_yaml.to_html(index=False, classes='table table-striped table-bordered')}
    </div>
</div>
"""
display(HTML(html_yaml))
```
